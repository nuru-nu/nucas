```python
import nucas
```

```python
nucas.utils.set_basedir_temp()
db = nucas.db.get_db()
```

```python
import mediapy

def train(target):
  config = nucas.train.get_config()
  config.target = f'https://www.robots.ox.ac.uk/~vgg/data/dtd/images/{target.split("_")[0]}/{target}.jpg'

  model, stats = nucas.train.train(config)
  imgs = nucas.run.run(config, model)
  mediapy.show_video(imgs)
  db.save(config, model, stats, imgs, overwrite=True)
```

```python
train('chequered_0045')
```

```python
train('blotchy_0060')
```

```python
train('bumpy_0107')
```
