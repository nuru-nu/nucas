```python
name = 'inception2_mixed4a_36__c0'
name = 'inception2_mixed4a_358__c0'

params = np.load(io.BytesIO(nucas.utils.fetch(f'https://nuru.nu/ncas/ncas/{name}.np')), allow_pickle=True)
```

```python
c = nucas.train.get_config()
m = nucas.backend.CaOrig(clip=2)
m.load_state_dict(nucas.utils.tf2pt(params))

imgs = nucas.run.run(c, m)
mediapy.show_video(imgs)
```
